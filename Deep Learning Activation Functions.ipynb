```python
import numpy as np
import matplotlib.pylab as plt
```

# Activation Function

```python
x=np.arange(-10,10,0.01)
print(len(x))
```

```python
y=[]
for i in x:
    if i>0:
        y.append(1)
    else:
        y.append(0)
print(len(y))
```

```python
plt.plot(x,y,color="red")
plt.title("Step Activation Function")
plt.xlabel("X")
plt.ylabel("Step(x)")
plt.grid()
plt.show()
```

# Linear Activation Function

```python
x=np.arange(-10,10,0.01)
y=x
```

```python
plt.plot(x,y,color="red")
plt.title("Linear Activation Function")
plt.xlabel("X")
plt.ylabel("Linear(x)")
plt.grid()
plt.show()
```

# Sigmoid Activation

```python
x=np.arange(-10,10,0.01)
y=1/(1+np.exp(-x))
```

```python
plt.plot(x,y,color="red")
plt.title("Sigmoid Activation Function")
plt.xlabel("X")
plt.ylabel("Sigmoid(x)")
plt.grid()
plt.show()
```

# Tanh Activation

```python
x=np.arange(-10,10,0.01)
y=np.tanh(x)
```

```python
plt.plot(x,y,color="red")
plt.title("Tanh Activation Function")
plt.xlabel("X")
plt.ylabel("Tanh()")
plt.grid()
plt.show()
```

# ReLU Activation

```python
x = np.arange(-10, 10, 0.01)
y = []
for i in x:
    if i > 0:
        y.append(i)
    else:
        y.append(0)
```

```python
plt.plot(x,y,color="red")
plt.title("ReLU Activation Function")
plt.xlabel("X")
plt.ylabel("ReLU(x)")
plt.grid()
plt.show()
```

# Leaky ReLU

```python
x = np.arange(-10, 10, 0.01)
y = []
for i in x:
    if i > 0:
        y.append(i)
    else:
        y.append(0.01 * i)
```

```python
plt.plot(x,y,color="Red")
plt.title("Leaky ReLU Activation Function")
plt.xlabel("X")
plt.ylabel("Leaky ReLU(x)")
plt.grid()
plt.show()
```

# Softmax Activation

```python
x = np.array([2.0, 1.0, 0.5])
exp_x = np.exp(x)
y = exp_x / np.sum(exp_x)
print(y)
print("Total:", np.sum(y))
```

```python
classes = ["Cat", "Dog", "Horse"]
plt.bar(classes, y)
plt.title("Softmax Output")
plt.xlabel("Classes")
plt.ylabel("Probability")
plt.show()
```
